# E-Commerce Sales & Customer Analytics

A hands-on notebook for data inspection, cleaning, relational joins, SQL analysis and visualization.

```
from pathlib import Path
import pandas as pd
ROOT = Path('..').resolve()
orders = pd.read_csv(ROOT/'data/sample/orders.csv', parse_dates=['order_purchase_timestamp','order_delivered_customer_date'])
items = pd.read_csv(ROOT/'data/sample/order_items.csv')
customers = pd.read_csv(ROOT/'data/sample/customers.csv')
products = pd.read_csv(ROOT/'data/sample/products.csv')
print(orders.shape, items.shape, customers.shape, products.shape)
```

```
orders.info()
orders.isna().sum().sort_values(ascending=False).head(10)
```

```
items = items.merge(products[['product_id','product_category_name']], on='product_id', how='left')
items['revenue'] = items['price'] + items['freight_value']
analysis = orders.merge(customers[['customer_id','customer_state','customer_city']], on='customer_id').merge(items[['order_id','product_category_name','revenue']], on='order_id')
analysis.head()
```

```
analysis[analysis['order_status']!='canceled'].groupby('customer_state')['revenue'].sum().sort_values(ascending=False).head(10)
```
